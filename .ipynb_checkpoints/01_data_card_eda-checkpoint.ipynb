````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from readiness_utils import (LABELS, RAW_FILE, FEATURE_FILE, NUM, CAT, find_data_dir,
                             to_tier, engineer_features)

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 50)
sns.set_theme(style="whitegrid")
TIER_COLORS = {"Green": "#2e9e5b", "Orange": "#f0a020", "Red": "#d64545"}

DATA_DIR = find_data_dir()
raw = pd.read_csv(DATA_DIR / RAW_FILE)
df = raw.copy()
print("Loaded", DATA_DIR / RAW_FILE, df.shape)
df.head()
````

## 1 · Structure: what is one row, and is the table complete?

````python
n_ath, n_days = df["Athlete_ID"].nunique(), df["Day"].max()
print("rows:", len(df), "| athletes:", n_ath, "| days per athlete:", n_days)
print("rows if every athlete had every day:", n_ath * n_days)
print("missing athlete-days:", n_ath * n_days - len(df))
print("duplicate (athlete, day) rows:", df.duplicated(["Athlete_ID", "Day"]).sum())
print()
print("Rows per athlete:")
print(df.groupby("Athlete_ID").size().value_counts().sort_index().to_string())
````

````python
# Which columns are fixed per athlete (profile) and which change every day?
for c in ["Age", "Gender", "Sport_Type"]:
    changes = (df.groupby("Athlete_ID")[c].nunique() > 1).sum()
    print(f"{c:12s} athletes whose value changes over time: {changes}")

pd.DataFrame({"dtype": df.dtypes.astype(str), "n_unique": df.nunique(),
              "example": df.iloc[0]})
````

**Things to note for the data card**
- One row = one athlete on one day. `Athlete_ID`, `Age`, `Gender` and `Sport_Type` are fixed per athlete, so rows from the same athlete are **not independent**. This is why every split in this project is done by athlete.
- `Stress_Level` is **Low / Medium / High** text, not a 1–10 number. We convert it to 1/2/3 for the models.
- The file name says *synthetic*. Say this in your limitations: results show that the pipeline works, not that it would be this accurate on real athletes.

## 2 · Missing values

````python
miss = df.isna().sum()
miss = miss[miss > 0].to_frame("missing")
miss["percent"] = (miss["missing"] / len(df) * 100).round(2)
miss
````

````python
# Is the missing sleep random, or does it follow a pattern?
df["Sleep_Missing"] = df["Sleep_Duration_Hours"].isna().astype(int)

print("Mean Recovery_Score  (sleep recorded vs missing):")
print(df.groupby("Sleep_Missing")["Recovery_Score"].mean().round(1).to_string())
print()
print("Share of rows missing sleep, by Training_Type:")
print(df.groupby("Training_Type")["Sleep_Missing"].mean().round(3).to_string())
print()
print("Athletes with the most missing sleep days:")
print(df.groupby("Athlete_ID")["Sleep_Missing"].sum().sort_values(ascending=False).head(5).to_string())

fig, ax = plt.subplots(figsize=(8, 3))
df.groupby("Day")["Sleep_Missing"].mean().plot(ax=ax, marker="o")
ax.set_title("Share of athletes with missing sleep, by day"); ax.set_ylabel("share")
plt.show()
````

**How we handle it (in `readiness_utils.engineer_features`):** a missing sleep value is replaced by that athlete's own average of **earlier** days, and a flag column `Sleep_Missing` keeps the fact that it was filled. If it is the athlete's first day, the pipeline fills with the median (inside the training folds only).

**Important:** the check above shows the missing-sleep days are *not random*: their average Recovery_Score is far lower than on days with sleep recorded. That is why we keep the `Sleep_Missing` flag. State this in your data card. Notebook 03 (section 4) tests how much the flag matters. In real use, a missing sleep reading would simply mean the device did not record it, so say that the flag is a property of this dataset.

Note the deck says ~600 missing values in *Sleep Duration and Training Intensity*. In this file, nearly all of them are Sleep (the check above shows the exact counts). Update the slide wording to match.

## 5 · Is a personal baseline worth using?

````python
# Share of variation that sits BETWEEN athletes (0 = athletes all alike, 1 = each athlete has a fixed level).
for col in ["HRV_ms", "Resting_Heart_Rate", "Sleep_Duration_Hours", "Recovery_Score"]:
    between = df.groupby("Athlete_ID")[col].mean().var()
    share = between / df[col].var()
    print(f"{col:22s} between-athlete share of variance: {share:.2f}")
````

If the share is high, comparing an athlete with **their own** usual level (which our `HRV_vs_Own` and `RHR_vs_Own` features and the driver analysis do) is useful. If it is low, athletes look alike and personal baselines add little. Write down which one you see.

## 6 · Leakage checks

````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GroupKFold, cross_val_score

y = df["Recovery_Score"]
groups = df["Athlete_ID"]
X_all = pd.get_dummies(df.drop(columns=["Recovery_Score", "Tier", "Athlete_ID"]), drop_first=True)
X_all = X_all.fillna(X_all.median())

def cv_r2(cols):
    rf = RandomForestRegressor(n_estimators=100, random_state=0, n_jobs=-1)
    return cross_val_score(rf, X_all[cols], y, cv=GroupKFold(5), groups=groups, scoring="r2").mean()

all_cols = list(X_all.columns)
no_subj = [c for c in all_cols if c not in ("Mood_Score", "Energy_Level")]
no_subj_cal = [c for c in no_subj if c not in ("Day", "Week", "Day_of_Week")]

print(f"R2, all columns:                            {cv_r2(all_cols):.3f}")
print(f"R2, without Mood_Score / Energy_Level:      {cv_r2(no_subj):.3f}")
print(f"R2, also without Day / Week / Day_of_Week:  {cv_r2(no_subj_cal):.3f}")
print(f"R2 from HRV_ms alone (correlation squared): {df['HRV_ms'].corr(y) ** 2:.3f}")
````

**How to read this**
- R² near **1.0** would mean `Recovery_Score` is just a formula of the inputs (leakage-like, so a model would look perfect for the wrong reason). A value clearly below 1 means there is real noise left, so the task is not trivial.
- `Recovery_Score` itself, and `Athlete_ID`, are never used as model inputs.
- `Mood_Score` and `Energy_Level` are self-reported scores. If dropping them barely changes R², leave them out (simpler, and no circular logic). That is the default in this project; `USE_SUBJECTIVE` in notebook 03 lets you test it.
- `HRV_ms` is by far the strongest single signal. Say so in the slides, and expect the models to rely on it heavily.

## 7 · Feature engineering and save

````python
import inspect
from readiness_utils import engineer_features
print(inspect.getsource(engineer_features))
````

| New feature | What it is | Why |
|---|---|---|
| `Stress_Ordinal` | Low / Medium / High → 1 / 2 / 3 | models need numbers |
| `Training_Load` | duration × intensity | your deck's definition |
| `Sleep_Filled`, `Sleep_Missing` | missing sleep filled from the athlete's earlier days, plus a flag | your imputation plan |
| `Sleep_Deficit_7d` | sum of (8 − sleep hours, floored at 0) over the last 7 days | your deck's rolling deficit |
| `Load_7d_Mean` | average training load over the last 7 days | accumulated load |
| `HRV_vs_Own`, `RHR_vs_Own` | today minus the athlete's earlier average | personal baseline |

All windows end **today** and never look at future days. (The 8-hour target is an assumption: state it in the data card.)

````python
features = engineer_features(raw)
features.to_csv(DATA_DIR / FEATURE_FILE, index=False)
print("Saved", DATA_DIR / FEATURE_FILE, features.shape)

print("\nMissing values left in the model inputs (filled inside the pipeline):")
left = features[NUM + CAT].isna().sum()
print(left[left > 0].to_string())
features[["Athlete_ID", "Day", "Training_Load", "Sleep_Filled", "Sleep_Deficit_7d",
          "Load_7d_Mean", "HRV_vs_Own", "RHR_vs_Own", "Recovery_Score"]].head(10)
````
